```python
import pandas as pd
import os

# Asegúrate de encontrarte en el directorio donde se encuentra el archivo.
df = pd.read_csv('/data_studentlist.csv', header='infer')
df.head(5)
```


1) Altura promedio de los estudiantes varones:

```python
altura_hombres = df[df['Gender'] == 'M']['Height'].mean()
print(f"Altura promedio de los varones: {altura_hombres:.2f}")
```

2) Altura promedio de las estudiantes:



```python
altura_estudiantes = df[df['Gender'] == 'F']['Height'].mean()
print(f"Altura promedio de las estudiantes: {altura_estudiantes:.2f}")
```

3) Peso promedio de los estudiantes varones:


```python
peso_hombres = df[df['Gender'] == 'M']['Weight'].mean()
print(f"Peso promedio de los varones: {peso_hombres:.2f}")
```

4) Peso promedio de las estudiantes:


```python
peso_mujeres = df[df['Gender'] == 'F']['Weight'].mean()
print(f"Peso promedio de las mujeres: {peso_mujeres:.2f}")
```

5) La mayor estatura entre los estudiantes varones:

```python
estatura_hombres = df[df['Gender'] == 'M']['Height'].max()
print(f"La mayor estatura entre los estudiantes varones: {estatura_hombres:.2f}")
```

6) La menor estatura entre las estudiantes:

```python
estatura_mujeres = df[df['Gender'] == 'M']['Height'].min()
print(f"La menor estatura entre las estudiantes mujeres: {estatura_mujeres:.2f}")
```

7) El peso más ligero entre los estudiantes varones de estatura superior a 1.75 m:

```python
peso_pluma = df[(df['Gender'] == 'M') & (df['Height'] > 175)]['Weight'].min()
print(f"El peso más ligero entre los varones de más de 1.75 m: {peso_pluma:.2f}")
```

8) El mayor peso entre las estudiantes de estatura inferior a 1.60 m:<

```python
peso_pesado = df[(df['Gender'] == 'F') & (df['Height'] < 160)]['Weight'].max()
print(f"El peso más pesado entre las mujeres de menos de 1.60 m: {peso_pesado:.2f}")
```

9) La nota promedio de los alumnos (hombres y mujeres) sin ausencias ('N'):

```python
promedio_n = df[df['Absence'] == 'N']['Grade'].mean()
print(f"La nota promedio de los alumnos sin ausencias: {promedio_n:.2f}")
```

10) La nota promedio de los alumnos (hombres y mujeres) con absentismo ('Y'):

```python
promedio_y = df[df['Absence'] == 'Y']['Grade'].mean()
print(f"La nota promedio de los aumnos con absentismo: {promedio_y:.2f}")
```

11) Altura promedio de los alumnos (hombres y mujeres) con grupo sanguíneo 'A' o 'AB':

```python
grupo_sanguineo = ['A', 'AB']
promedio = df[df['Bloodtype'].isin(grupo_sanguineo)]['Height'].mean()
print(f"La altura promedio de los alumnos con grupo sanguíneo A o AB: {promedio:.2f}")
```

12) Altura promedio de los alumnos varones con grupo sanguíneo 'A' o 'AB':

```python
grupo_sanguineo = ['A', 'AB']
promedio_h = df[(df['Gender'] == 'M') & (df['Bloodtype'].isin(grupo_sanguineo))]['Height'].mean()
print(f"La altura promedio de los varones con grupo sanguíneo A o AB: {promedio_h:.2f}")
```

13) Edad promedio de los alumnos (hombres y mujeres) con absentismo ('Y') cuya nota es mayor o igual a 3:

```python
edad_promedio = df[(df['Absence'] == 'Y') & (df['Grade'] >= 3)]['Age'].mean()
print(f"Edad promedio de los alumnos con absentismo y nota >= 3: {edad_promedio:.2f}")
```

```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
%matplotlib inline

df = pd.read_csv('/data_coffeeshop.csv', header='infer', na_values=[' '])
print(df.shape)
df.head(5)
```

1. Genera una tabla de frecuencias para la variable 'yearOfStart' y visualízala por año.
•	Ordena por año.
•	Dibuja una gráfica de línea de 1997 a 2014. <= Pista: plt.xlim()


```python
df.columns = df.columns.str.strip()
# 1. Tabla de frecuencias ordenada por año
frecuencias = df['Year of Start'].value_counts().sort_index()
print(frecuencias)

# 2. Dibujar la gráfica de línea
frecuencias.plot(kind='line', marker='o', color='b')

# 3. Limitar la vista de 1997 a 2014
plt.xlim(1997, 2014)

# 4. (Opcional) Ponerle títulos para que se entienda mejor
plt.title("Apertura de negocios por año")
plt.xlabel("Año")
plt.ylabel("Cantidad")

# Mostrar la gráfica
plt.show()
```

2.  Divide los datos por el estado actual del negocio ('In' o 'Out'). Después, visualiza la tendencia annual de las frecuencias de la variable 'yearOfStart'.
•	Ordena por año.
•	Dibuja dos gráficas de líneas superpuestas de 1997 a 2014.
•	Usa el objeto 'figure'.


```python
# 1. Crear grafica
plt.figure(figsize=(10, 6))
# 2. negocios que siguen abiertos ('In')
frecuencias_in = df[df['Current State'] == 'In']['Year of Start'].value_counts().sort_index()
# 3. negocios que ya cerraron ('Out')
frecuencias_out = df[df['Current State'] == 'Out']['Year of Start'].value_counts().sort_index()
# 4. Negocios abiertos = verde, negocios cerrados = rojo
frecuencias_in.plot(kind='line', marker='o', color='green', label='In (Siguen abiertos)')
frecuencias_out.plot(kind='line', marker='x', color='red', label='Out (Ya cerraron)')
# 5. Limitar 1997 a 2014
plt.xlim(1997, 2014)
# 6. Estetica grafica
plt.title("Aperturas por año: Negocios Abiertos vs Cerrados")
plt.xlabel("Año")
plt.ylabel("Cantidad de negocios")
plt.legend()
# Mostrar la gráfica
plt.show()
```

**Respuesta:** La tendencia muestra un gran crecimiento en la creacion de negoocios a partir del 2008. Lo cual demuestra que los negocios que abrieron antes de ese año tienen mayor tendencia a que cierren por lo pesado que es mantener un negocio cuando crece la compotencia.
